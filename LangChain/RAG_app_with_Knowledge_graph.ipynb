Create an entity retirever

```python
#now we gonna check on, how we can collect the nodes and relationships match from the database  when we give data,
# first we create a entity retiever which we create a prompt that can understand the entites in the given data input
from langchain_core.prompts import ChatPromptTemplate
entity_prompt=ChatPromptTemplate.from_messages(
    [
        (
            "system","You are extracting entities from the text",
        ),
        (
            "human","Use the following information to extract entities"
            "input:{question}",
        ),
    ]
)
```

```python
"""
No we create a class to store the enitties appears in the data
for that we use a special type of class called Pydantic model. This class is used to define the structure of data you expect. We want an object called Entities, and it must contain one field called names.
names must be a list of strings.
Entities class is inherited from the BaseModel.  
"""
from pydantic import BaseModel,Field
from typing import List

class Entities(BaseModel):
    names:List[str] = Field(
        ...,
        description="All the entities that appear in the text",
    )
```

```python
#create the chain
entity_chain=(entity_prompt|llm.with_structured_output(Entities))
```

```python
#check the functinality of created chain, we check whtether the chain can identify the entities in a given detail. Output confirms that it has identified the entities X and Y and give them as a list as per Entities class. 
entity_chain.invoke({"question" "who learn the both x and Y subjects"}).names
```

Create a graph Retirever

```python
#Now we going to build graph retriever. This means we are going to identify the relationships from the database for entities in the input data.
#first we remove the lucene charachters from the input data. 
#Then 
from langchain_neo4j.vectorstores.neo4j_vector import remove_lucene_chars

#This function converts a plain text input string into a fuzzy full-text search query. 
"""
Sanitizes input — remove_lucene_chars() strips special Lucene characters (like +, -, &&, !, (, ), etc.) that would break the query parser.
Tokenizes — splits the input string into individual words.
Appends ~2 to each word — this is a fuzzy match operator (edit distance of 2), meaning it matches words within 2 character substitutions/insertions/deletions of the original.
Joins words with AND — all terms must be present in the matched document.
"""
def generate_full_text_query(input):
    full_text_query=""
    words=[el for el in remove_lucene_chars(input).split() if el]
    for word in words[:-1]:
        full_text_query+=f"{word}~2 AND"
    full_text_query+=f"{words[-1]}~2"
    return full_text_query.strip()
```

```python
"""
Now we define the function for graph retirever. 
First we identify the entities in the input data using the previously prepared entity retirever.
Then we mathc the relationships from the databse for these identified entities. For that we use a cypher query on our graph database
"""
def graph_retriever(question: str) -> str:
    
    # Create fulltext index if it doesn't exist
    graph.query(
        "CREATE FULLTEXT INDEX entity IF NOT EXISTS FOR (n:__Entity__) ON EACH [n.id]"
    )
    
    result = ""
    entities = entity_chain.invoke({"question": question})
    
    for entity in entities.names:
        response = graph.query(
            """CALL db.index.fulltext.queryNodes('entity', $query, {limit:2})
            YIELD node, score
            CALL {
              WITH node
              MATCH (node)-[r:!MENTIONS]->(neighbor)
              RETURN node.id + ' - ' + type(r) + ' -> ' + neighbor.id AS output
              UNION ALL
              WITH node
              MATCH (node)<-[r:!MENTIONS]-(neighbor)
              RETURN neighbor.id + ' - ' + type(r) + ' -> ' + node.id AS output
            }
            RETURN output LIMIT 20
            """,
            {"query": generate_full_text_query(entity)},
        )
        result += "\n".join([el['output'] for el in response])
    
    return result
```

Sementic Retriever using a hybrid search with vector database and the graph database

```python
"""
we can use neo4jvector to use the vector search. so we need a embedding method to apply on our chunks.
First we create the vector index , Then apply similarity search
""" 
from langchain_neo4j import Neo4jVector

vector_index = Neo4jVector.from_existing_graph(
    embedding,#previous embedding
    graph=graph,#created graph
    search_type="hybrid",
    node_label="Document",
    text_node_properties=["text"],
    embedding_node_property="embedding"
)
```

```python
#now check the functionality of semetic retirever with a simple input, we retrieve first 2 most matching chunks from the text (alrdy introduce text data) by having k=2. 
vector_index.similarity_search("who learn both X and Y subjects", k=2)
```

Create a hybrid retirever with both graph retirever and sementic retirever and then use it in a chain

```python
def retriever(question):
    graph_search_result = graph_retriever(question)
    semantic_search_result = [data.page_content for data in vector_index.similarity_search(question, k=2)]
    final_data = f"Graph data:{graph_search_result}\nText data:{' '. join(semantic_search_result)}"
    return final_data
```

```python
#define prompt tempalte for RAG
from langchain_core.prompts import ChatPromptTemplate
prompt = ChatPromptTemplate.from_messages(
    [
        (
            "system","Answer this question using the provided context only.",
        ),
        (
            "human","Context: {context}"
            "Question: {question}",
        ),
    ]
)
```

```python
#create RAG chain
from langchain_core.runnables import RunnablePassthrough
new_chain = (
    {
      "context": retriever,
      "question": RunnablePassthrough()
    }
    | prompt
    | llm
)
   
```

```python
#invoke the chian with and example question
response = new_chain.invoke("who learn both X and Y subjects")
print(response.content)
```
